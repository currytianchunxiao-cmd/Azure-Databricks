```sql
alter table dev_ext.dwi_ext.emp add column active_flg string;
```

```sql
-- Merge for update existing records and insert new records
merge into dev_ext.dwi_ext.emp t1
using dev_ext.sdi_ext.emp t2
on t1.id = t2.id
when matched then
update set
t1.salary = t2.salary,
t1.active_flg = 'Y'
when not matched then 
insert 
(id, name, job, salary, active_flg)
values (t2.id, t2.name, t2.job, t2.salary, 'Y')
-- when not matched by source and t1.active_flg = 'Y' then
when not matched by source then
update set
active_flg = 'N'
;
```
