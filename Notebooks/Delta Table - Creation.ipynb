# TEMP VIEW & PERMANENT VIEW

```sql
create temporary view v_emp as select * from dev_ext.sdi_ext.emp
where salary > 80000;
```

```sql
select * from v_emp
```

```sql
create view dev_ext.sdi_ext.v_emp as select * from dev_ext.sdi_ext.emp
where salary > 80000;
```

```sql
select * from dev_ext.sdi_ext.v_emp;
```

```sql
create or replace view dev_ext.sdi_ext.v_emp as select * from dev_ext.sdi_ext.emp
where salary > 90000;
select * from dev_ext.sdi_ext.v_emp;
```
