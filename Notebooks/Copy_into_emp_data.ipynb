```python
%sql
--create table with specifical columns
create or replace table dev_ext.bronze.emp_cp_col
(
  employee_id	int,
  department_id	int,
  name	string,
  age	int,
  create_ts timestamp
);
```

```python
%sql
--loading data to dev_ext.bronze.emp_cp_col
COPY INTO dev_ext.bronze.emp_cp_col
FROM (
  select cast(employee_id as int) employee_id, cast(department_id as int) department_id,
  name, cast(age as int) age, current_timestamp() create_ts
  from
  "/Volumes/dev_ext/sdi_ext/vol_ext/input"
)
FILEFORMAT = CSV
PATTERN = '*.csv'
FORMAT_OPTIONS (
  'mergeSchema' = 'true',
  'header' = 'true'
)
COPY_OPTIONS (
  'mergeSchema' = 'true'
);
```
