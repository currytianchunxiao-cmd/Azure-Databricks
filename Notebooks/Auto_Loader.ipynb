# addNewColumns (Default when schema is not provided): Automatically adds new columns to the schema as they are discovered. When a change is detected, the stream fails with an UnknownFieldException and must be restarted (typically handled by a Databricks Job retry) to incorporate the new columns.

```python
df = (
    spark
    .readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("pathGlobFilter", "*.csv")
    .option("header", "true")
    .option("cloudFiles.schemaHints", "employee_id int, department_id int, age int")
    .option("cloudFiles.schemaLocation", "/Volumes/dev_ext/bronze/loading/checkping/autoloader/1/")
    .load("/Volumes/dev_ext/bronze/loading/autoloader_input/*/")
)
```

```python
# loading new data to delta table - dev_ext.bronze.emp_al_1
from pyspark.sql.functions import col
(
    df
    .withColumn("__file", col("_metadata.file_name"))
    .writeStream
    .option("mergeSchema", "true")
    .option("checkpointLocation", "/Volumes/dev_ext/bronze/loading/checkping/autoloader/1/")
    .outputMode("append")
    .trigger(availableNow=True)
    .toTable("dev_ext.bronze.emp_al_1")
)
```
