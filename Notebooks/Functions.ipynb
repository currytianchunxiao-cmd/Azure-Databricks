```sql
-- CREATE SCALAR FUNCTION TO CALCULATE TAX SQL
CREATE OR REPLACE FUNCTION dev_ext.bronze.tax_sql(salary double)
RETURNS DOUBLE
LANGUAGE SQL
RETURN salary * 0.3;
```

```sql
select dev_ext.bronze.tax_sql(1000) as tax
```

```sql
select *, dev_ext.bronze.tax_sql(salary) as tax from dev_ext.dwi_ext.emp;
```

```sql
%python
from pyspark.sql.functions import expr

# Read the employee table from the Unity Catalog
df = spark.read.table("dev_ext.dwi_ext.emp")

# Use the SQL scalar function `tax_sql` to calculate tax on the salary column
df.withColumn("tax", expr("dev_ext.bronze.tax_sql(salary)")).display()
```

```sql
-- Improved version with error handling
CREATE OR REPLACE FUNCTION dev_ext.bronze.fruit_nutrition_safe(fruit_name STRING)
RETURNS STRING
LANGUAGE PYTHON
AS
$$
import requests
try:
    api_url = f"https://www.fruityvice.com/api/fruit/{fruit_name}"
    response = requests.get(api_url, timeout=5)
    response.raise_for_status() # Raise error for HTTP codes like 404/500
    data = response.json()
    return str(data.get('nutritions', 'NA'))
except Exception as e:
    return f"Error: {str(e)}"
$$;
```

```sql
select dev_ext.bronze.fruit_nutrition_safe("apple")
```

```sql
-- CREATE TABLE FUNCTION TO GET ALL EMP FROM A PARTICULAR DEPT
CREATE OR REPLACE FUNCTION dev_ext.bronze.get_emp(emp_name string)
RETURNS TABLE(emp_name string, position string)
LANGUAGE SQL
RETURN (
    SELECT name, job 
    FROM dev_ext.dwi_ext.emp
    WHERE name = emp_name
);
```

```sql
select * from dev_ext.bronze.get_emp("Alice")
```
