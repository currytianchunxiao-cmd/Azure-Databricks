```sql
create schema if not exists sdi;
```

```sql
create schema dwi;
```

```sql
CREATE OR REPLACE TEMP VIEW employee_sample AS
SELECT * FROM VALUES
  (1, 'Alice', 'Data Engineer', 85000),
  (2, 'Bob', 'Data Scientist', 95000),
  (3, 'Charlie', 'BI Developer', 75000),
  (4, 'David', 'Spark Developer', 90000)
AS employee(id, name, job, salary);
```

```sql
select * from employee_sample;
```

```sql
-- Create MANAGED table
CREATE TABLE IF NOT EXISTS sdi.employee_managed
COMMENT 'Unity Catalog Managed Table'
AS SELECT * FROM employee_sample;
```

```sql
use catalog bigdata_datalake_databricks;
show tables dropped in sdi;
```

```sql
undrop table sdi.employee_managed;
```

```sql
select * from sdi.employee_managed;
```

```sql
-- Verify
SELECT * FROM sdi.employee_managed;
```

```sql
-- Check table type
DESCRIBE TABLE EXTENDED sdi.employee_managed;
```

# Azure ADLS Gen2

```sql
-- Create EXTERNAL table
CREATE TABLE sdi.employee_external
COMMENT 'Unity Catalog External Table on ADLS Gen2'
LOCATION 'abfss://bigdata4dev@azuredls.dfs.core.windows.net/poc/data/data/employee'
AS SELECT * FROM employee_sample;
```

```sql
-- Create EXTERNAL table using Databricks Unity Catalog DEFAULT STORAGE
CREATE TABLE sdi.employee_external
COMMENT 'Unity Catalog External Table (using Databricks default storage)'
LOCATION 'abfss://unity-catalog-storage@dbstorage4lfilolqqo3su.dfs.core.windows.net/7405608601959635/external/employee_external'
AS SELECT * FROM employee_sample;
```

```sql
%sql
-- List storage details for the managed table
DESCRIBE DETAIL sdi.employee_managed;
```

```sql
-- Verify
SELECT * FROM sdi.employee_external;
```

```sql
-- Check table type
DESCRIBE TABLE EXTENDED sdi.employee_external;
```

```sql
DROP TABLE sdi.employee_managed;

-- Data files are gone forever
```

```sql
-- List storage details for the managed table
DESCRIBE DETAIL sdi.employee_managed;
```

```sql
DROP TABLE sdi.employee_external;

-- ONLY metadata is deleted
-- Data files still exist in ADLS Gen2
```

```sql
DESCRIBE DETAIL sdi.employee_external;
```

```sql
CREATE TABLE sdi.employee_external_v2
COMMENT 'Unity Catalog External Table (using Databricks default storage)'
LOCATION 'abfss://unity-catalog-storage@dbstorage4lfilolqqo3su.dfs.core.windows.net/7405608601959635/external/employee_external'
```

```sql
-- Data is back
SELECT * FROM sdi.employee_external_v2;
```
