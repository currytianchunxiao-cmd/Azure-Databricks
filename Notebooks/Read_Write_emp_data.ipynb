```python
df = spark.read.option("header","true").csv("dbfs:/Volumes/dev_ext/sdi_ext/vol_ext/files/emp.csv")
df.show()
```

```python
dbutils.widgets.text('dept_id', '', 'Deparment ID')
```

```python
_dept = dbutils.widgets.get('dept_id')
print(_dept)
```

```python
df_filter = df.where(df.department_id == _dept)
df_filter.show()
```

```python
_count = df_filter.count()
if _count > 0:
    df_filter.write.mode("overwrite").saveAsTable(f"dev_ext.sdi.emp_{_dept}")
    print(f"Inserted {_count} records into dev_ext.sdi.emp_{_dept}")
else:
    print(f"No records found for dept_id {_dept}")
```
