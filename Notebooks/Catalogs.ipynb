```sql
create catalog dev;
```

```sql
-- EXTERNAL LOCATION
CREATE EXTERNAL LOCATION `ext_catalog_v2`
URL 'abfss://data@azuredls.dfs.core.windows.net/adb/catalog_ext'
WITH (STORAGE CREDENTIAL `bigdata_databricks_credentials`)
;
```

```sql
-- CREATE CATALOG WITH EXTERNAL LOCATION FOR MANAGED TABLES
CREATE CATALOG dev_ext
MANAGED LOCATION 'abfss://data@azuredls.dfs.core.windows.net/adb/catalog'
COMMENT 'This is an external storage catalog'
;
```

```sql
-- CREATE CATALOG WITH EXTERNAL LOCATION FOR MANAGED TABLES
CREATE CATALOG dev_ext_v2
MANAGED LOCATION 'abfss://data@azuredls.dfs.core.windows.net/adb/catalog_ext'
COMMENT 'This is an external storage catalog version 2'
;
```

```sql
describe catalog extended dev_ext;
```
