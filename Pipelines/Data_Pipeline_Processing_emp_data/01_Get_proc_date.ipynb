```python
dbutils.widgets.text("proc_date","")
```

```python
_proc_date = dbutils.widgets.get("proc_date")
print(_proc_date)
```

```python
spark.conf.set("spark.sql.legacy.timeParsePolicy","LEGACY")

_input_date = spark.sql(f"""
                       select date_format(to_timestamp('{_proc_date}','yyyy-MM-dd'), 'E')
                       """).collect()[0][0]
print(_input_date)
```

```python
dbutils.jobs.taskValues.set(key="proc_date", value=_input_date)
```
